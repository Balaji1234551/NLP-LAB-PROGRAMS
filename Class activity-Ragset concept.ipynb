### Explanation of NLTK Penn Treebank Tags:
* **`PRP`**: Personal Pronoun (e.g., "I")
* **`VBP`**: Verb, non-3rd person singular present (e.g., "LOVE")
* **`NNP`**: Proper Noun, singular (e.g., "INDIA")

```python
import spacy

# Load the small English pipeline
nlp = spacy.load("en_core_web_sm")

doc = nlp("I LOVE INDIA")

print(f"{'Word':<10} | {'Universal Tag':<15} | {'Detailed Tag (Penn)':<20} | {'Description'}")
print("-" * 70)
for token in doc:
    print(f"{token.text:<10} | {token.pos_:<15} | {token.tag_:<20} | {spacy.explain(token.tag_)}")
```
