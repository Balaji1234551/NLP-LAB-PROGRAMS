```python
import re
text = "My name is Balaji. My phone number is 9876543210. Email: balaji@gmail.com"
# Search for a word
result = re.search("Balaji", text)
if result:
    print("Word found:", result.group())
else:
    print("Word not found")
# Match a pattern at the beginning
result = re.match("My", text)
if result:
    print("Pattern matched:", result.group())
else:
    print("Pattern not matched")
# Find all numbers
numbers = re.findall(r"\d+", text)
print("Numbers found:", numbers)
# Find email address
email = re.findall(r"[\w.-]+@[\w.-]+\.\w+", text)
print("Email address:", email)
```
