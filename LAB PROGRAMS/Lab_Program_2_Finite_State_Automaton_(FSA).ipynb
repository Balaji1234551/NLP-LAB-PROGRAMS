```python
def finite_automaton(s):
    state = 0
    for ch in s:
        if state == 0:
            if ch == 'a':
                state = 1
            else:
                state = 0

        elif state == 1:
            if ch == 'a':
                state = 1
            elif ch == 'b':
                state = 2
            else:
                state = 0

        elif state == 2:
            if ch == 'a':
                state = 1
            else:
                state = 0

    if state == 2:
        print("String Accepted")
    else:
        print("String Rejected")


s = input("Enter a string containing a and b: ")
finite_automaton(s)
```
