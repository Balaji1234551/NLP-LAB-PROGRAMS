```python
text = [
    ("I", "PRON"),
    ("eat", "VERB"),
    ("rice", "NOUN"),
    ("I", "PRON"),
    ("like", "VERB"),
    ("rice", "NOUN")
]

sentence = ["I", "like", "rice"]

tags = {}

for word, tag in text:
    if word not in tags:
        tags[word] = {}
    tags[word][tag] = tags[word].get(tag, 0) + 1

print("POS Tags:")

for word in sentence:
    best_tag = max(tags[word], key=tags[word].get)
    print(word, "->", best_tag)
```
