```python
import nltk
from nltk.tokenize import word_tokenize
from nltk import RegexpTagger

nltk.download('punkt')
nltk.download('punkt_tab')

text = "The cat runs quickly"

words = word_tokenize(text)

patterns = [
    (r'^The$', 'DET'),
    (r'.*ly$', 'ADV'),
    (r'.*ing$', 'VERB'),
    (r'.*s$', 'NOUN'),
    (r'.*', 'NOUN')
]

tagger = RegexpTagger(patterns)

print(tagger.tag(words))
```
