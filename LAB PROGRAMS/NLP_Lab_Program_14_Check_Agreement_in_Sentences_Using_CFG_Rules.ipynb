```python

from nltk import CFG
from nltk.parse import ChartParser

grammar = CFG.fromstring("""
S -> NP_SG VP_SG
S -> NP_PL VP_PL

NP_SG -> 'he' | 'she' | 'the boy'
NP_PL -> 'they' | 'we' | 'the boys'

VP_SG -> 'runs' | 'plays' | 'eat'
VP_PL -> 'run' | 'play' | 'eat'
""")

parser = ChartParser(grammar)

sentence = input("Enter a sentence: ").lower().split()

# Join multiword subjects into grammar-compatible tokens
if sentence[:2] == ['the', 'boy']:
    sentence = ['the boy'] + sentence[2:]
elif sentence[:2] == ['the', 'boys']:
    sentence = ['the boys'] + sentence[2:]

trees = list(parser.parse(sentence))

if trees:
    print("Sentence has valid agreement")
    for tree in trees:
        print(tree)
else:
    print("Sentence has invalid agreement")
```
