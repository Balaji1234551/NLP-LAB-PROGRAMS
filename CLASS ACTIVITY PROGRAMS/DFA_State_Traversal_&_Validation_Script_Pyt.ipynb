```python
class DFAEndingWithAB:
    def __init__(self):

        self.states = {"Q0", "Q1", "Q2"}
        self.start_state = "Q0"
        self.accept_states = {"Q2"}

    def run(self, input_string):
        current_state = self.start_state
        path = [current_state]

        print(f"--- Processing Input: '{input_string}' ---")
        print(f"Start State: {current_state}")

        for char in input_string:
            if char not in {'a', 'b'}:
                print(f"Error: Invalid character '{char}' in input string.")
                return False

            # Transition Logic
            if current_state == "Q0":
                if char == 'a':
                    current_state = "Q1"
                else:  # char == 'b'
                    current_state = "Q0"

            elif current_state == "Q1":
                if char == 'a':
                    current_state = "Q1"
                else:  # char == 'b'
                    current_state = "Q2"

            elif current_state == "Q2":
                if char == 'a':
                    current_state = "Q1"
                else:  # char == 'b'
                    current_state = "Q0"

            path.append(current_state)
            print(f"Read '{char}' -> Transitioned to state {current_state}")

        is_accepted = current_state in self.accept_states
        print(f"Path of States: {' -> '.join(path)}")
        if is_accepted:
            print("Result: Accepted\n")
        else:
            print("Result: Rejected\n")

        return is_accepted

dfa = DFAEndingWithAB()
test_strings = ["ab", "aab", "baba", "aba", "bb", "a", "b"]

for s in test_strings:
    dfa.run(s)
```
